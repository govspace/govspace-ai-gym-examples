# 🧩 Interactive Widgets Demo

This notebook shows how to make interactive plots using **ipywidgets**.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

x = np.linspace(0, 10, 200)
```

## 🔄 Interactive sine wave
Use the slider to adjust frequency.

```python
def plot_sine(freq=1.0):
    plt.figure(figsize=(6,3))
    plt.plot(x, np.sin(2 * np.pi * freq * x))
    plt.title(f"Sine wave, freq={freq} Hz")
    plt.xlabel("x")
    plt.ylabel("sin(2πfx)")
    plt.ylim(-1.1, 1.1)
    plt.show()

interact(plot_sine, freq=(0.1, 5.0, 0.1))
```

## 🎨 Try changing the plot style
You can swap `plt.plot` with markers or colors, or adjust the `interact` ranges.
