# 🏛️ Interactive Supervised Entities Dashboard

This notebook lets you explore **entity compliance and incidents** interactively.
Filter by **Jurisdiction** or **Entity Type** using dropdowns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ipywidgets import interact, Dropdown

sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("data/supervised_entities_demo.csv")
```

## 🔄 Interactive Filter and Plots

```python
def interactive_dashboard(jurisdiction='All', entity_type='All'):
    filtered = df.copy()
    if jurisdiction != 'All':
        filtered = filtered[filtered['Jurisdiction'] == jurisdiction]
    if entity_type != 'All':
        filtered = filtered[filtered['Entity_Type'] == entity_type]
    
    fig, axes = plt.subplots(1, 2, figsize=(12,5))
    
    # Compliance Score by Entity
    sns.barplot(data=filtered.sort_values('Compliance_Score', ascending=False), 
                x='Compliance_Score', y='Entity_Name', palette='Greens_r', ax=axes[0])
    axes[0].set_title('Compliance Score by Entity')
    axes[0].set_xlabel('Compliance Score')
    axes[0].set_ylabel('Entity')
    
    # Incidents vs Compliance
    sns.scatterplot(data=filtered, x='Incidents', y='Compliance_Score', hue='Entity_Type', s=100, ax=axes[1])
    axes[1].set_title('Incidents vs Compliance')
    axes[1].set_xlabel('Incidents')
    axes[1].set_ylabel('Compliance Score')
    axes[1].legend(bbox_to_anchor=(1.05,1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

## 🛠️ Widgets

```python
jur_options = ['All'] + sorted(df['Jurisdiction'].unique().tolist())
type_options = ['All'] + sorted(df['Entity_Type'].unique().tolist())

interact(interactive_dashboard, 
         jurisdiction=Dropdown(options=jur_options, description='Jurisdiction:'),
         entity_type=Dropdown(options=type_options, description='Entity Type:'))
```

## 🎉 Usage
- Select a **Jurisdiction** to filter entities by region.
- Select an **Entity Type** to focus on Agencies or Contractors.
- Both plots update dynamically.
