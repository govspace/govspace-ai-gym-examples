# 🌐 Mini Dashboard (Plotly Express)

This notebook shows interactive charts using **Plotly Express** with a small built-in dataset.

```python
import plotly.express as px
```

## 📊 Load dataset
We'll use the `gapminder` dataset (small subset).

```python
df = px.data.gapminder()
df.head()
```

## 🌍 Scatter plot: GDP vs Life Expectancy

```python
fig = px.scatter(df.query("year==2007"), 
                 x="gdpPercap", y="lifeExp",
                 size="pop", color="continent",
                 hover_name="country", size_max=60,
                 title="GDP per Capita vs Life Expectancy (2007)")
fig.show()
```

## 📈 Line chart: Life expectancy over time

```python
fig = px.line(df.query("country=='Canada'"), 
              x='year', y='lifeExp',
              title='Canada Life Expectancy over Time')
fig.show()
```
