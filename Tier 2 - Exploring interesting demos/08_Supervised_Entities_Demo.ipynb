# 🏛️ Supervised Entities Dashboard Demo

Analyze and visualize the performance and compliance of supervised entities (agencies and contractors).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## 📥 Load dataset

```python
df = pd.read_csv("data/supervised_entities_demo.csv")
df
```

## 📊 Compliance Score by Entity

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df.sort_values('Compliance_Score', ascending=False), 
            x='Compliance_Score', y='Entity_Name', palette='Greens_r')
plt.title('Compliance Score by Entity')
plt.xlabel('Compliance Score')
plt.ylabel('Entity')
plt.show()
```

## ⏱️ Incidents vs Compliance

```python
sns.scatterplot(data=df, x='Incidents', y='Compliance_Score', hue='Entity_Type', s=100)
plt.title('Incidents vs Compliance Score')
plt.xlabel('Incidents')
plt.ylabel('Compliance Score')
plt.legend(bbox_to_anchor=(1.05,1), loc='upper left')
plt.show()
```

## 📊 Average Compliance by Jurisdiction

```python
jur_avg = df.groupby('Jurisdiction')['Compliance_Score'].mean().reset_index()
sns.barplot(data=jur_avg, x='Compliance_Score', y='Jurisdiction', palette='Blues_r')
plt.title('Average Compliance by Jurisdiction')
plt.xlabel('Average Compliance Score')
plt.ylabel('Jurisdiction')
plt.show()
```

## 🎉 Done!

This lightweight demo allows GovTech users to explore **entity performance, compliance, and incidents** interactively.
