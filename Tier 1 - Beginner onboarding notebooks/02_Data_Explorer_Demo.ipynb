# 📊 Data Explorer Demo

This notebook demonstrates how to load, explore, and visualize a dataset using **pandas** and **seaborn**.

We'll use the built-in `tips` dataset (no files needed!).

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

## 📥 Load dataset
Seaborn provides some great small datasets for demos.

````python
df = sns.load_dataset("tips")
df.head()
````

## 🔍 Quick summary

````python
df.describe()
````

## 📈 Scatter plot
Visualize the relationship between **total bill** and **tip**.

````python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="time")
plt.title("Tips: Total Bill vs Tip Size")
plt.show()
````

## 📊 Bar chart (average tips by day)

````python
avg = df.groupby("day")["tip"].mean().reset_index()
sns.barplot(data=avg, x="day", y="tip")
plt.title("Average Tip by Day")
plt.show()
````
