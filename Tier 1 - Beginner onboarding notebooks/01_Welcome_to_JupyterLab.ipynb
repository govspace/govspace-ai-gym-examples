> ⚠️ **Before you edit any notebook — please read this first**
>
> The `TEMPLATES/` folder is **read-only**. It is refreshed from GitHub every time you start your server, so anything you try to save inside `TEMPLATES/` will fail with a *"Permission denied"* error (and would be overwritten on the next restart anyway).
>
> **To work on any exercise:**
>
> 1. In the JupyterLab file browser on the left, **right-click the notebook → Duplicate**.
> 2. **Drag the duplicate out of `TEMPLATES/`** into your home folder (one level up, alongside the `TEMPLATES/` folder).
> 3. Open and edit *that* copy — it's writable and won't be overwritten on restart.
>
> The same applies if you want to create a new notebook from scratch: create it in your home folder, **not** inside `TEMPLATES/`.
