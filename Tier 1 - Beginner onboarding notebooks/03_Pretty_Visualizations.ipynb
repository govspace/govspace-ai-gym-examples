# 🎨 Pretty Visualizations

This notebook demonstrates several attractive visualizations using **seaborn** and **matplotlib**.

All examples use tiny built-in datasets.

````python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
````

## 🌡️ Heatmap
A color-coded matrix is great for showing correlations.

````python
df = sns.load_dataset("penguins").dropna()
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Penguin Feature Correlations")
plt.show()
````

## 📈 Scatterplot with regression line

````python
sns.lmplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.title("Bill Dimensions by Species")
plt.show()
````

## 📊 Histogram + KDE
Shows a smooth estimation of the distribution.

````python
sns.histplot(df["flipper_length_mm"], kde=True, bins=20, color="purple")
plt.title("Distribution of Flipper Lengths")
plt.show()
````
